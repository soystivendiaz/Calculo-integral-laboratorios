```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
```

```python
f = x
g = x**2

area = sp.integrate(f - g, (x, 0, 1))

print("Area exacta:")
print(area)

print("Area decimal:")
print(float(area))
```

```python
def f_num(x):
    return x

def g_num(x):
    return x**2

x_vals = np.linspace(0, 1, 300)

plt.plot(x_vals, f_num(x_vals), label='f(x)=x')
plt.plot(x_vals, g_num(x_vals), label='g(x)=x^2')
plt.fill_between(x_vals, f_num(x_vals), g_num(x_vals), alpha=0.4)

plt.xlabel('x')
plt.ylabel('y')
plt.title('Area entre dos curvas')
plt.grid(True)
plt.legend()
plt.show()
```

```python
R = sp.sqrt(x)
r = x

V_arandelas = sp.pi * sp.integrate(R**2 - r**2, (x, 0, 1))

print("Volumen por arandelas:")
print(V_arandelas)

print("Valor decimal:")
print(float(V_arandelas))
```

```python
f = sp.sqrt(x)
g = x

V_capas = 2*sp.pi*sp.integrate(x*(f - g), (x, 0, 1))

print("Volumen por capas:")
print(V_capas)

print("Valor decimal:")
print(float(V_capas))
```
