```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

n = sp.symbols('n', positive=True, integer=True)
x = sp.symbols('x')
```

```python
a_n = n/(n + 1)

terminos = [a_n.subs(n, k) for k in range(1, 11)]

print(terminos)
```

```python
limite = sp.limit(a_n, n, sp.oo)

print(limite)
```

```python
serie = sp.summation((sp.Rational(1, 2))**n, (n, 1, sp.oo))

print(serie)
```

```python
sumas = []

for N in range(1, 11):
    S_N = sum([(1/2)**k for k in range(1, N+1)])
    sumas.append(S_N)

print(sumas)
```

```python
N = np.arange(1, 11)
sumas = np.array([sum([(1/2)**k for k in range(1, m+1)]) for m in N])

plt.plot(N, sumas, marker='o')
plt.axhline(1, linestyle='--')
plt.xlabel('N')
plt.ylabel('S_N')
plt.title('Sumas parciales de la serie geometrica')
plt.grid(True)
plt.show()
```

```python
taylor = sp.series(sp.exp(x), x, 0, 6)

print(taylor)
```
