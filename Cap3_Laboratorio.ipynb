```python
import sympy as sp

x = sp.symbols('x')
```

```python
def verificar(f, F):
    return sp.simplify(sp.diff(F, x) - f)
```

```python
f = x*sp.exp(x)
F = sp.integrate(f, x)

print(F)
print(verificar(f, F))
```

```python
f = sp.sqrt(9 - x**2)
F = sp.integrate(f, x)

print(F)
print(verificar(f, F))
```

```python
f = (3*x + 5)/(x**2 + 3*x + 2)

descomposicion = sp.apart(f, x)
F = sp.integrate(f, x)

print(descomposicion)
print(F)
print(verificar(f, F))
```
