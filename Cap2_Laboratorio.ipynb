```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')
```

```python
f = 3*x**2 + 2*x + 1
F = sp.integrate(f, x)

print("Primitiva encontrada:")
print(F)
```

```python
verificacion = sp.diff(F, x)

print("Derivada de la primitiva:")
print(verificacion)
```

```python
print(sp.simplify(sp.diff(F, x) - f))
```

```python
f = 2*x*sp.cos(x**2)
F = sp.integrate(f, x)

print("Primitiva encontrada:")
print(F)
```

```python
print("Verificacion:")
print(sp.diff(F, x))
```

```python
u = x**2 + 1
du = sp.diff(u, x)

print("u =", u)
print("du/dx =", du)
```

```python
f = 2*x/(x**2 + 1)
F = sp.integrate(f, x)

print("Primitiva encontrada:")
print(F)
```

```python
def f_num(x):
    return x**2

a = 0
b = 2
n = 4

dx = (b - a) / n

print("Delta x =", dx)
```

```python
x_izq = np.linspace(a, b - dx, n)
suma_izquierda = np.sum(f_num(x_izq) * dx)

print("Puntos izquierdos:", x_izq)
print("Suma izquierda:", suma_izquierda)
```

```python
x_der = np.linspace(a + dx, b, n)
suma_derecha = np.sum(f_num(x_der) * dx)

print("Puntos derechos:", x_der)
print("Suma derecha:", suma_derecha)
```

```python
x_medios = np.linspace(a + dx/2, b - dx/2, n)
suma_medios = np.sum(f_num(x_medios) * dx)

print("Puntos medios:", x_medios)
print("Suma de punto medio:", suma_medios)
```

```python
x_vals = np.linspace(a, b, 300)
y_vals = f_num(x_vals)

plt.plot(x_vals, y_vals, label='f(x)=x^2')
plt.bar(x_izq, f_num(x_izq), width=dx, align='edge',
        alpha=0.4, edgecolor='black')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Aproximacion del area con suma izquierda')
plt.grid(True)
plt.legend()
plt.show()
```

```python
def f_num(x):
    return x**2

x_vals = np.linspace(0, 2, 300)
y_vals = f_num(x_vals)

plt.plot(x_vals, y_vals, label='f(x)=x^2')
plt.fill_between(x_vals, y_vals, alpha=0.4)

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Area bajo la curva de f(x)=x^2 en [0,2]')
plt.grid(True)
plt.legend()
plt.show()
```

```python
t = sp.symbols('t')

A = sp.integrate(t**2, (t, 0, x))

print("Funcion de acumulacion:")
print(A)

print("Derivada de la funcion de acumulacion:")
print(sp.diff(A, x))
```

```python
t = sp.symbols('t')

v = t**2 - 4*t + 3

print("Velocidad:")
print(v)
```

```python
desplazamiento = sp.integrate(v, (t, 0, 4))

print("Desplazamiento:")
print(desplazamiento)
print(float(desplazamiento))
```

```python
ceros = sp.solve(v, t)

print("Ceros de la velocidad:")
print(ceros)
```

```python
d1 = sp.integrate(v, (t, 0, 1))
d2 = sp.integrate(v, (t, 1, 3))
d3 = sp.integrate(v, (t, 3, 4))

distancia = abs(d1) + abs(d2) + abs(d3)

print("Distancia recorrida:")
print(distancia)
print(float(distancia))
```

```python
def v_num(t):
    return t**2 - 4*t + 3

t_vals = np.linspace(0, 4, 400)
v_vals = v_num(t_vals)

plt.plot(t_vals, v_vals, label='v(t)=t^2-4t+3')
plt.axhline(0, color='black', linewidth=1)

plt.fill_between(t_vals, v_vals, 0, where=(v_vals >= 0),
                 alpha=0.3, label='Velocidad positiva')

plt.fill_between(t_vals, v_vals, 0, where=(v_vals < 0),
                 alpha=0.3, label='Velocidad negativa')

plt.xlabel('t')
plt.ylabel('v(t)')
plt.title('Velocidad en el intervalo [0,4]')
plt.grid(True)
plt.legend()
plt.show()
```
